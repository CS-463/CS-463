## Part 1: Vectors

A data point with $p$ features is a list of $p$ numbers, a **vector**: $\mathbf{x} = (x_1, x_2, \dots, x_p)$. A customer with a balance of \$800 and an income of \$40,000 is the vector $(800, 40000)$. With two features, you can draw a vector two ways: as a **point** at $(x_1, x_2)$, or as an **arrow** from the origin to that point. Both pictures are useful, and we'll switch between them.

**Rows and columns.** In matrix math, a vector is written as a **column** by convention: $\mathbf{x} = \begin{bmatrix} x_1 \\ x_2 \end{bmatrix}$. To save space, we'll often write it across the page as $(x_1, x_2)$; it's the same vector. (NumPy's 1D arrays, like `np.array([1, 2])`, are neither a row nor a column.)

Three operations, each done entry by entry:

- **Adding:** $(2, 1) + (-1, 3) = (1, 4)$. Drawn as arrows: put the tail of the second vector at the tip of the first. The sum is the vector from the tail of the first to the tip of the second.
- **Scaling** by a number: $2 \times (1.5, 1) = (3, 2)$. A positive number stretches or shrinks the vector without turning it. A negative number also flips it around.
- **Subtracting** is adding the negative: $\mathbf{q} - \mathbf{p} = \mathbf{q} + (-\mathbf{p})$, where $-\mathbf{p} = (-1)\mathbf{p}$ has the same length as $\mathbf{p}$ but points the opposite way. For example, $(3, 2) - (1, 1) = (3, 2) + (-1, -1) = (2, 1)$. So it's drawn tip to tail, like any sum (the second code cell below draws it: the left panel shows $-\mathbf{p}$, and the middle panel shows the sum).

**The fact to remember: $\mathbf{q} - \mathbf{p}$ is the vector from $\mathbf{p}$ to $\mathbf{q}$.** Check it: start at $\mathbf{p}$, add $\mathbf{q} - \mathbf{p}$, and you land on $\mathbf{q}$. So drawn as an arrow starting at the tip of $\mathbf{p}$, it ends at the tip of $\mathbf{q}$, and subtracting two points gives the vector that goes from one to the other. The right panel above draws it. Part 4 uses this.

**Length.** The length of a vector comes from the Pythagorean theorem:

$$
\lVert \mathbf{x} \rVert = \sqrt{x_1^2 + x_2^2 + \dots + x_p^2}.
$$

For example, $\lVert (3, 4) \rVert = \sqrt{9 + 16} = 5$. Scaling multiplies the length: $\lVert 2\mathbf{x} \rVert = 2\lVert \mathbf{x} \rVert$. The **distance between two points** $\mathbf{p}$ and $\mathbf{q}$ is the length of the vector between them, $\lVert \mathbf{q} - \mathbf{p} \rVert$.

A **unit vector** is any vector with length 1. Dividing a vector by its length gives the unit vector pointing in the same direction: $(3, 4)/5 = (0.6, 0.8)$.

<details>
<summary>Confirming (0.6, 0.8) is a unit vector</summary>

$$
\lVert (0.6, 0.8) \rVert = \sqrt{0.6^2 + 0.8^2} = \sqrt{0.36 + 0.64} = \sqrt{1} = 1.
$$

</details>

We can rewrite that original vector, $(3, 4)$ as $5\times(0.6,0.8)$. The same procedure could be applied to any vector, so every vector is a length times a direction (a unit vector).

**Code worth reading.** In NumPy, `+`, `-`, and `*` on arrays work **elementwise** (entry by entry). We use them for vector addition, subtraction, and scaling. `np.linalg.norm` computes the length.

````python
u = np.array([2, 1]) # some vector we're calling u
v = np.array([-1, 3]) # some vector we're calling v
print('u + v  =', u + v) # the sum of u and v
print('v - u  =', v - u) # add the negative of u to v ("subtract" u from v)
print('2 * v  =', 2 * v) # scale v by 2 (keep its direction, double its length)
print('length of (3, 4):', np.linalg.norm(np.array([3, 4]))) # compute the magnitude (length) of the vector (3, 4)
print('unit vector:     ', np.array([3, 4]) / np.linalg.norm(np.array([3, 4]))) # compute the unit vector in the direction of (3, 4)

# Make sure you understand what the denominator of that last line is!
````

## Part 2: The dot product

<details>
<summary>Why dot products matter (optional).</summary>

The dot product (defined just below) is one of the most common operations in machine learning. A few places it shows up:

- **Linear models.** Every prediction from linear regression, and every log-odds from logistic regression, is a dot product of a coefficient vector with a feature vector, plus an intercept. You'll see how below.
- **Matrix multiplication is many dot products.** Multiplying a matrix by a vector computes one dot product per row of the matrix (Parts 6–8). Neural networks, starting Monday, are built from these, and GPUs are designed to compute enormous numbers of them in parallel.
- **Similarity.** Various measures based on dot products measure a notion of "similarity". For example, represent each document as a vector of word counts. Two articles on the same topic point in similar directions, even if one is ten times longer than the other.
- **Embeddings.** Later in the course we'll study **embeddings**: vectors that represent words, sentences, or images in a special space where similar meanings point in similar directions. There, the dot product (or cosine similarity) of two embeddings measures how similar the *concepts themselves* are: the embedding of "puppy" has a high cosine similarity with "dog" and a low one with "spreadsheet". Search engines, recommendation systems, and chatbots that look up documents all rank their results this way.
- **QKV Circuits.** Transformers, the models behind modern LLMs, use dot products to decide which words in a passage to pay **attention** to. Each word's embedding is mapped, by a matrix as in Part 7, to three vectors: a **query** (what this word is looking for), a **key** (what this word offers), and a **value** (the information it passes along). The dot product of one word's query with another word's key scores how relevant the second word is to the first: the same similarity idea as with embeddings. The scores become weights, and each word's new representation is the weighted sum of the values.

</details>

**The algebraic definition.** Multiply matching entries, then add:

$$
\mathbf{w} \cdot \mathbf{x} = w_1 x_1 + w_2 x_2 + \dots + w_p x_p.
$$

The result is a single number. For $\mathbf{w} = (4, 2)$ and $\mathbf{x} = (3, 9)$: $\mathbf{w}\cdot\mathbf{x} = w_1x_1 + w_2x_2 = 4 \cdot 3 + 2 \cdot 9 = 12 + 18 = 30$.

**You've used this many times.** A dot product is a **weighted sum** of the features. Linear regression's prediction, and logistic regression's log-odds (what the Chapter 4 notebook called the *evidence score*), here is log-odds again:

$$
\log(\text{odds}) = \beta_0 + \beta_1 x_1 + \dots + \beta_p x_p.
$$

If we collect the $\beta$ coefficients into a vector, $\boldsymbol{\beta}$, and the $x_i$ into another vector, $\mathbf{x}$, then we can rewrite the log odds as a dot product (plus $\beta_0$):

$$
\beta_0 + \beta_1 x_1 + \dots + \beta_p x_p = \beta_0 + \boldsymbol{\beta}\cdot\mathbf{x}.
$$

**Neural Network Conventions:**

For most of this notebook we'll adopt the conventions of neural networks, and write the coefficients as $\mathbf{w}$ and the intercept as $b$. $\mathbf{w}$ for **weights**, $b$ for **bias** (don't confuse this with the "bias" of "bias-variance" tradeoff). So $\mathbf{w}\cdot\mathbf{x} + b$ is the same thing as $\boldsymbol{\beta}\cdot\mathbf{x} + \beta_0$.

**The geometric formula.** The dot product is also

$$
\mathbf{w} \cdot \mathbf{x} = \lVert \mathbf{w} \rVert \, \lVert \mathbf{x} \rVert \cos\theta,
$$

where $\theta$ is the angle between the two vectors.

<details>
<summary>Optional: checking the example with the geometric formula</summary>

**A vector dotted with itself.** Replace $\mathbf{w}$ with $\mathbf{x}$ in the geometric formula. The angle between $\mathbf{x}$ and itself is $0$, and $\cos 0 = 1$, so

$$
\mathbf{x}\cdot\mathbf{x} = \lVert \mathbf{x} \rVert\,\lVert \mathbf{x} \rVert \cos 0 = \lVert \mathbf{x} \rVert^2,
\qquad \text{so} \qquad
\lVert \mathbf{x} \rVert = \sqrt{\mathbf{x}\cdot\mathbf{x}}.
$$

By the algebraic definition, $\mathbf{x}\cdot\mathbf{x} = x_1^2 + x_2^2 + \dots + x_p^2$, so this is the length formula from Part 1.

**Checking the example.** For $\mathbf{w} = (4, 2)$ and $\mathbf{x} = (3, 9)$, the lengths are

$$
\lVert \mathbf{w} \rVert = \sqrt{\mathbf{w}\cdot\mathbf{w}} = \sqrt{4^2 + 2^2} = \sqrt{20},
\qquad
\lVert \mathbf{x} \rVert = \sqrt{\mathbf{x}\cdot\mathbf{x}} = \sqrt{3^2 + 9^2} = \sqrt{90}.
$$

The angle between them is $\theta = 45°$, and $\cos 45° = 1/\sqrt{2}$. Plugging these into the geometric formula,

$$
\mathbf{w}\cdot\mathbf{x} = \lVert \mathbf{w} \rVert\,\lVert \mathbf{x} \rVert \cos\theta = \sqrt{20}\,\sqrt{90} \cdot \frac{1}{\sqrt{2}} = \sqrt{\frac{1800}{2}} = \sqrt{900} = 30,
$$

the same answer as the algebraic definition.

</details>

**Sign and size.** Lengths are never negative, so the sign of $\mathbf{w}\cdot\mathbf{x}$ comes from $\cos\theta$ alone. Keep the two lengths fixed and rotate $\mathbf{x}$: only $\cos\theta$ changes, and it's between $-1$ and $1$, so the dot product has a maximum and minimum possible value if you allow $\theta$ to vary and hold the vector lengths fixed:

$$
-\lVert \mathbf{w} \rVert\,\lVert \mathbf{x} \rVert \;\le\; \mathbf{w}\cdot\mathbf{x} \;\le\; \lVert \mathbf{w} \rVert\,\lVert \mathbf{x} \rVert.
$$

- The dot product is **largest** when $\mathbf{x}$ points exactly the same way as $\mathbf{w}$ ($\theta = 0°$, $\cos\theta = 1$).
- The dot product is **positive** while the angle is between  0° and 90°, shrinking as $\mathbf{x}$ rotates toward perpendicular, and it's **0** at $\theta = 90°$.
- With $90° \lt \theta \lt 180°$ the dot product is **negative**, and magnitude is largest ("most negative") when $\mathbf{x}$ points exactly the opposite way ($\theta = 180°$).

So the dot product measures **how closely $\mathbf{x}$ lines up with $\mathbf{w}$**, scaled by both lengths. Making either vector longer scales the dot product up, but never changes its sign.

**The perpendicular test.** Two vectors are perpendicular exactly when their dot product is 0. The rest of the lab uses this test constantly.

**The shadow.** Shine a light straight down onto the line through $\mathbf{w}$, so $\mathbf{x}$ casts a shadow on it. The shadow's signed length is

$$
\frac{\mathbf{w}\cdot\mathbf{x}}{\lVert \mathbf{w} \rVert} = \lVert \mathbf{x} \rVert \cos\theta,
$$

positive if the shadow falls ahead of the origin (on $\mathbf{w}$'s side) and negative if behind. So **the dot product measures how far $\mathbf{x}$ reaches in $\mathbf{w}$'s direction**, scaled by the length of $\mathbf{w}$. In the example, the shadow is $30/\sqrt{20} \approx 6.71$.

**Code worth reading.** `w @ x` computes the dot product of two 1D arrays. The angle comes from the geometric formula, solved for $\cos\theta$.

````python
w = np.array([4, 2])
x = np.array([3, 9])
print('w @ x   =', w @ x)

# The dot product, spelled out: multiply matching entries (w[0] with x[0], w[1] with x[1]), then add.
print(f'by hand = w[0]*x[0] + w[1]*x[1] = {w[0]}*{x[0]} + {w[1]}*{x[1]} =', w[0] * x[0] + w[1] * x[1])

cos_theta = (w @ x) / (np.linalg.norm(w) * np.linalg.norm(x))
print('angle   =', np.degrees(np.arccos(cos_theta)).round(1), 'degrees')
````

**All of them.** Do the same with an unknown $x_1$: $3x_1 + x_2 = 0$ gives $x_2 = -3x_1$. So every perpendicular vector has the form $(x_1, -3x_1) = x_1(1, -3)$: a **multiple of $(1, -3)$**. Both examples fit, with $x_1 = 1$ and $x_1 = -2$. Drawn as points, these vectors make a **line through the origin**, at a right angle to $(3, 1)$.

</details>

### Step 2: why it's a line, perpendicular to $\mathbf{w}$

By Part 2's perpendicular test, $\mathbf{w}\cdot\mathbf{x} = 0$ says that the vector $\mathbf{x}$ is **at a right angle to $\mathbf{w}$**. So the set of all solutions, $\mathbf{x}$, is *every vector at a right angle to $\mathbf{w}$*. **Reading the figure.**

- **Blue vectors (left):** solutions of $\mathbf{w}\cdot\mathbf{x} = 0$. Each one is at a right angle to $\mathbf{w}$, and together they all lie along one line through the origin, whichever way they point and however long they are.
- **Faded gray vectors:** two vectors that are **not** solutions, labeled with their dot products with $\mathbf{w}$. $(1, 1)$ is less than 90° from $\mathbf{w}$, so $\mathbf{w}\cdot\mathbf{x} = 3 > 0$. $(-1, 0)$ is more than 90° away, so $\mathbf{w}\cdot\mathbf{x} = -2 < 0$. Neither lies on the line: only vectors at **exactly** 90° to $\mathbf{w}$ do.

With three features, the same set (of $\mathbf{x}$) is a **plane** through the origin perpendicular to $\mathbf{w}$, as in the right panel. With $p$ features it's a flat $(p-1)$-dimensional surface called a **hyperplane**. Our pictures stay in 2D.

### Step 3: $\mathbf{w}\cdot\mathbf{x} = c$ for any $c$

Now let the right-hand side be any number $c$. With $\mathbf{w} = (2, 1)$, solve for $x_2$:

$$
\begin{aligned}
\mathbf{w}\cdot\mathbf{x} &= c \\
(2, 1)\cdot(x_1, x_2) &= c \\
2x_1 + x_2 &= c \\
x_2 &= -2x_1 + c.
\end{aligned}
$$

That's a line in slope-intercept form, with slope $-2$ and $x_2$-intercept $c$. For $c = 0$, it's Step 2's line through the origin. Changing $c$ changes only the intercept, so:

- **Lines for different values of $c$ are parallel.** They all have slope $-2$: changing $c$ slides the line without rotating it.
- **Every line of this form is perpendicular to $\mathbf{w}$.** They're all parallel, so they're perpendicular to the same vector.

The same rearrangement works for any $\mathbf{w}$ whose second entry isn't 0. There are vector-algebra proofs of these facts that work with any number of features (and for vertical lines), but for our purposes, this will suffice.

<details>
<summary>Optional: a vector proof that works with any number of features</summary>

**Two points on the line.** Pick any two points $\mathbf{p}$ and $\mathbf{q}$ on the line $\mathbf{w}\cdot\mathbf{x} = c$. "On the line" means each point satisfies the line's equation. So, putting each point in place of $\mathbf{x}$,

$$
\mathbf{w}\cdot\mathbf{p} = c \qquad\text{and}\qquad \mathbf{w}\cdot\mathbf{q} = c.
$$

**The vector between them.** By Part 1, $\mathbf{q} - \mathbf{p}$ is the vector from $\mathbf{p}$ to $\mathbf{q}$. Both of its ends are on the line, so it runs along the line.

**Dot it with $\mathbf{w}$.** Written out with two features (more features just add more terms):

$$
\begin{aligned}
\mathbf{w}\cdot(\mathbf{q} - \mathbf{p}) &= w_1(q_1 - p_1) + w_2(q_2 - p_2) && \text{algebraic definition of the dot product} \\
&= (w_1 q_1 + w_2 q_2) - (w_1 p_1 + w_2 p_2) && \text{multiply out; group the } q \text{ terms and the } p \text{ terms} \\
&= \mathbf{w}\cdot\mathbf{q} - \mathbf{w}\cdot\mathbf{p} && \text{each group is a dot product} \\
&= c - c && \text{both points are on the line: } \mathbf{w}\cdot\mathbf{q} = c \text{ and } \mathbf{w}\cdot\mathbf{p} = c \\
&= 0 && \text{look where we started (top, left side of first equation)}.
\end{aligned}
$$

So by the perpendicular test (Part 2), $\mathbf{q} - \mathbf{p}$ is perpendicular to $\mathbf{w}$. That holds for **every** pair of points on the line, so the line runs perpendicular to $\mathbf{w}$. Nothing here needed two features or $w_2 \neq 0$. And since every line $\mathbf{w}\cdot\mathbf{x} = c$ is perpendicular to the same $\mathbf{w}$, they're all parallel.

**Example.** With $\mathbf{w} = (2, 1)$ and $c = 4$, the points $\mathbf{p} = (2, 0)$ and $\mathbf{q} = (1, 2)$ are both on the line: $2(2) + 0 = 4$ and $2(1) + 2 = 4$. The vector between them is $\mathbf{q} - \mathbf{p} = (-1, 2)$, and $\mathbf{w}\cdot(\mathbf{q} - \mathbf{p}) = 2(-1) + 1(2) = 0$.

</details>


### Step 4: where the line sits

Which of the parallel lines is which? Here are the results you'll need, stated without proof. The optional section below explains them.

- **Bigger $c$ puts the line farther out in the direction $\mathbf{w}$ points.**
- **Negative $c$ puts it on the other side of the origin.**
- **$c = 0$ puts it through the origin.**

The figure shows the lines for $c = -4, -2, 0, 2, 4$ with $\mathbf{w} = (2, 1)$. The dots mark where each line crosses $\mathbf{w}$'s direction.

<details>
<summary>Optional: why these results hold</summary>

Walk from the origin in $\mathbf{w}$'s direction: the points $t\mathbf{w}$, for numbers $t$. The line $\mathbf{w}\cdot\mathbf{x} = c$ crosses that path where

$$
\mathbf{w}\cdot(t\mathbf{w}) = t\,(\mathbf{w}\cdot\mathbf{w}) = t\,\lVert \mathbf{w} \rVert^2 = c, \qquad \text{so} \qquad t = \frac{c}{\lVert \mathbf{w} \rVert^2}.
$$

- **Where the crossing point is.** It's $t\mathbf{w} = \dfrac{c}{\lVert \mathbf{w} \rVert^2}\,\mathbf{w}$, at distance $\lvert t \rvert\,\lVert \mathbf{w} \rVert = \lvert c \rvert / \lVert \mathbf{w} \rVert$ from the origin.
- **Why it's the closest point.** The line crosses $\mathbf{w}$'s direction at a right angle (Step 3), and the shortest path from a point to a line is the one that meets it at a right angle.
- **The first three results.** $\lVert \mathbf{w} \rVert^2 > 0$, so $t$ has the sign of $c$ and grows with $c$. A bigger $c$ means a bigger $t$: farther out in $\mathbf{w}$'s direction. A negative $c$ means a negative $t$: the other side of the origin. And $c = 0$ means $t = 0$: the origin itself.
- **The spacing.** Raising $c$ by 1 raises the distance $c/\lVert \mathbf{w} \rVert$ by $1/\lVert \mathbf{w} \rVert$, so the lines for $c$ values one unit apart are $1/\lVert \mathbf{w} \rVert$ apart.

**Example.** With $\mathbf{w} = (2, 1)$ and $c = 4$: $\lVert \mathbf{w} \rVert^2 = 5$, so the crossing point is $\tfrac{4}{5}(2, 1) = (1.6, 0.8)$, at distance $4/\sqrt5 \approx 1.79$ from the origin. In the figure, neighboring lines are $2/\sqrt5 \approx 0.89$ apart, since their $c$ values differ by 2.

</details>

## Part 4: Adding $b$: sliding the line, and which side has a positive "score"

### Step 1: $b$ is just Part 3, rearranged

Throughout machine learning, we typically don't work with $\mathbf{w}\cdot\mathbf{x} = c$, instead, subtract c from both sides and replace $-c$ with $b$:

$$
\mathbf{w}\cdot\mathbf{x} + b = 0.
$$

That's Part 3's line with $c = -b$. So everything from Part 3 applies:

- It's **perpendicular to $\mathbf{w}$.**
- Changing $b$ **slides the whole line along $\mathbf{w}$'s direction**, without changing the slope, just as changing $c$ did in Part 3.

**Concrete.** With $\mathbf{w} = (1, 1)$, the values $b = 0, -1, -2, -3$ give the lines $x_1 + x_2 = 0, 1, 2, 3$, shown below.

````python
plot_level_lines(w=(1, 1), values=[0, -1, -2, -3], kind='b');
````

**Changing $b$ slides the line along $\mathbf{w}$'s direction without turning it.** Which way? This is the step people most often get backwards, so slowly:

- The line is $\mathbf{w}\cdot\mathbf{x} = -b$.
- **Increasing $b$ decreases $-b$.** 
- So **increasing $b$ moves the line opposite to the way $\mathbf{w}$ points.** 

### Step 2: the score, and where it is positive

We'll call $\mathbf{w}\cdot\mathbf{x} + b$ the **score** of $\mathbf{x}$. Starting Monday, it's written $z$. (The Chapter 4 notebook already wrote $z$ for the log-odds $\beta_0 + \beta_1 X$, which has exactly this form.) The score is 0 on the line. What is it off the line?

**The simplest case, $b = 0$.** The line $\mathbf{w}\cdot\mathbf{x} = 0$ passes through the origin, and the score is just $\mathbf{w}\cdot\mathbf{x}$. Part 2's sign result answers the question:

- A point on the side $\mathbf{w}$ points toward is **less than 90°** from $\mathbf{w}$, so its score is **positive**.
- A point on the other side is **more than 90°** from $\mathbf{w}$, so its score is **negative**.
- A point on the line is exactly 90° from $\mathbf{w}$, so its score is 0.

**Any $b$.** A similar argument, measuring angles from a point $\mathbf{x}_0$ on the line instead of from the origin, gives the same result. The figure shows both cases.

````python
plot_score_sides();
````

**So the line splits the plane in two, and $\mathbf{w}$ points toward the positive side.**

<details>
<summary>Optional: why the general case works</summary>

Pick any point $\mathbf{x}_0$ on the line, so $\mathbf{w}\cdot\mathbf{x}_0 + b = 0$, which means $b = -\mathbf{w}\cdot\mathbf{x}_0$. Substitute that into the score:

$$
\mathbf{w}\cdot\mathbf{x} + b = \mathbf{w}\cdot\mathbf{x} - \mathbf{w}\cdot\mathbf{x}_0 = \mathbf{w}\cdot(\mathbf{x} - \mathbf{x}_0).
$$

The last step uses the fact that the dot product distributes over subtraction (Part 3's optional proof spells this out). So the score of $\mathbf{x}$ is $\mathbf{w}$ dotted with the vector from $\mathbf{x}_0$ to $\mathbf{x}$: the gray vectors in the right panel. Part 2's sign result applies to that dot product exactly as in the $b = 0$ case: positive when $\mathbf{x}$ is on $\mathbf{w}$'s side of the line, negative on the other side, and 0 on the line.

</details>

### Step 3: logistic regression, seen this way

In the Chapter 4 notebooks, logistic regression modeled the log-odds as a linear combination of the features. (We write the predicted probability as $\hat p$ here, since $p$ already means the number of features.)

$$
\ln\!\left(\frac{\hat p}{1-\hat p}\right) = \beta_0 + \beta_1 X_1 + \beta_2 X_2 + \cdots + \beta_p X_p.
$$

Everything after $\beta_0$ is a sum of products, coefficient times feature, which is exactly Part 2's algebraic definition of a dot product. Collect the coefficients into a vector $\boldsymbol{\beta} = (\beta_1, \dots, \beta_p)$ and the features into a vector $\mathbf{x} = (X_1, \dots, X_p)$. Then

$$
\beta_1 X_1 + \beta_2 X_2 + \cdots + \beta_p X_p = \boldsymbol{\beta}\cdot\mathbf{x},
\qquad\text{so}\qquad
\ln\!\left(\frac{\hat p}{1-\hat p}\right) = \beta_0 + \boldsymbol{\beta}\cdot\mathbf{x}.
$$

That's a score $\mathbf{w}\cdot\mathbf{x} + b$, with $\mathbf{w} = \boldsymbol{\beta}$ and $b = \beta_0$. (The intercept $\beta_0$ stays outside the dot product, just like $b$.) Two consequences:

- **Its decision boundary is a line of this kind.** It predicts class 1 when $\hat p > 0.5$, which is exactly when the log-odds is positive. So the boundary is the line where the log-odds is 0, and it's perpendicular to $\boldsymbol{\beta}$.
- **$\boldsymbol{\beta}$ points toward the class-1 side.** By Step 2, $\boldsymbol{\beta}$ points toward the side where the log-odds is positive, which is where $\hat p > 0.5$.

The figure fits a logistic regression to two features (with scikit-learn) and draws the boundary and $\boldsymbol{\beta}$'s direction.

## Part 6: Many points at once: `X @ w + b` *(reading only)*

Neural networks are usually represented in terms of matrix multiplications. We're going to start with a matrix times a vector and build up to products of two matrices.

A **matrix** is a rectangular grid of numbers. A matrix with $n$ rows and $p$ columns is called $n \times p$, and the entry in row $i$ and column $j$ is written $X_{ij}$. A data set with $n$ points and $p$ features is an $n \times p$ matrix $X$: **one row per point** (its $p$ feature values) and one column per feature, exactly as in every data frame you've used.

**The row view of a matrix-vector product.** Multiplying $X$ by $\mathbf{w}$ dots **every row** of $X$ with $\mathbf{w}$:

$$
X\mathbf{w} = \begin{pmatrix} \text{row}_1 \cdot \mathbf{w} \\ \text{row}_2 \cdot \mathbf{w} \\ \vdots \\ \text{row}_n \cdot \mathbf{w} \end{pmatrix}.
$$

For example, with three points and two features:

$$
X = \begin{bmatrix} 1 & 2 \\ 3 & -1 \\ 0 & 4 \end{bmatrix}, \quad
\mathbf{w} = \begin{bmatrix} 2 \\ 1 \end{bmatrix}, \qquad
X\mathbf{w} = \begin{bmatrix} (1)(2) + (2)(1) \\ (3)(2) + (-1)(1) \\ (0)(2) + (4)(1) \end{bmatrix} = \begin{bmatrix} 4 \\ 5 \\ 4 \end{bmatrix}.
$$

Each entry is one point (one row) dotted with $\mathbf{w}$. That's exactly what your loop in 4(b) computed, one point at a time; $X\mathbf{w}$ computes them all at once.

**Shapes.** $X$ is a matrix with dimensions $n \times p$. $\mathbf{w}$ is a vector. You can think of $\mathbf{w}$ as a matrix with one column, so its dimensions are $p \times 1$. The rule for any matrix product:

$$
\underset{m \times k}{A}\;\;\underset{k \times r}{B} \;=\; \underset{m \times r}{AB}
$$

- **The inner dimensions must match** (the two $k$'s in the middle). Each row of $A$ is dotted with each column of $B$, so they need the same number of entries. If the inner dimensions differ, the product isn't defined.
- **The outer dimensions give the result's shape:** $m \times r$.

In the example above, $X$ was $3 \times 2$ and $\mathbf{w}$ was $2 \times 1$, so $X\mathbf{w}$ was $3 \times 1$. In general, for the scores:

$$
\underset{n \times p}{X}\;\;\underset{p \times 1}{\mathbf{w}} \;=\; \underset{n \times 1}{X\mathbf{w}}
$$

The inner dimensions match because each point (row) has $p$ features and $\mathbf{w}$ has one weight per feature. The result is $n \times 1$: one score per row (point). (NumPy stores `w` as a 1D array of shape `(p,)` rather than `(p, 1)`, and `X @ w` returns shape `(n,)`, but the rule is the same.)

*Notice the layouts:* $X$ stores each point as a **row**, and $\mathbf{w}$ is a **column**, so $X\mathbf{w}$ dots each row with $\mathbf{w}$. In Part 7, we'll stack several weight vectors as the **rows** of a matrix $W$ (instead of as columns), the convention used in math and in PyTorch. Part 8 shows how `.T` reconciles that with $X$'s rows.

**Adding $b$:** `X @ w + b` adds the single number `b` to every entry. That's all of the points' scores in one line.

Two more steps from logistic regression, both entry by entry:

- **Probabilities from scores:** `1 / (1 + np.exp(-scores))` applies the sigmoid $\sigma(z) = 1/(1 + e^{-z})$ to every score, giving each point's predicted probability of class 1.
- **Classes from probabilities:** `(probs > threshold).astype(int)` is 1 where the predicted probability is above the **decision threshold**, and 0 elsewhere. The usual threshold is 0.5. (At 0.5, this is the same as checking which side of the line a point is on, since $\sigma(0) = 0.5$.) Changing the threshold tunes the classifier: a lower threshold predicts class 1 for more points, and a higher one for fewer.

**Code worth reading.** Five points scored against the line $\mathbf{w}\cdot\mathbf{x} + b = 0$, with $\mathbf{w} = (3, 2)$ and $b = -6$.

````python
X_demo = np.array([[3, 0],
                   [0, 0],
                   [1, 1],
                   [2, 2],
                   [-1, 4]])
w_demo = np.array([3, 2])                               # w
b_demo = -6                                             # b

scores_demo = X_demo @ w_demo + b_demo                  # one score per row (point)
probs_demo = 1 / (1 + np.exp(-scores_demo))             # the sigmoid of every score

threshold = 0.5                                         # the decision threshold, in probability
classes_demo = (probs_demo > threshold).astype(int)     # 1 where the probability is above it
classes_low = (probs_demo > 0.25).astype(int)           # a lower threshold: more points get class 1

print('point      score  probability  class (0.5)  class (0.25)')
for point, s, pr, k, k_low in zip(X_demo, scores_demo, probs_demo, classes_demo, classes_low):
    print(f'{str(point):10} {s:5}  {pr:11.3f}  {k:11}  {k_low:12}')
````

## Part 7: A matrix maps one space to another

So far, $\mathbf{w}$ and $b$ have turned each point into one number, its score. This part looks at the same computation a different way: as a **map**, a rule that sends every point of one space to a point of another space. Understanding neural networks requires this view.

### Step 1: one row, from the plane to a number line

The points $(x_1, x_2)$ live in the **feature space**: the plane. The score $z = \mathbf{w}\cdot\mathbf{x} + b$ sends each of them to a single number, so it maps the whole plane onto a number line, a one-dimensional **z-space**. Written with matrices, $\mathbf{w}$ is now laid out as a **row**, a $1 \times 2$ matrix, not the column of Part 6. That's how weight matrices store weights: one row per weight vector (Step 2).

$$
\underset{1 \times 2}{\begin{bmatrix} w_1 & w_2 \end{bmatrix}}\;
\underset{2 \times 1}{\begin{bmatrix} x_1 \\ x_2 \end{bmatrix}} + b
\;=\; \underset{1 \times 1}{z}.
$$

Where does each point land? Points on the same level line (Part 3) have the same score, so **a whole level line lands on a single spot** of the number line. Points farther along $\mathbf{w}$'s direction land farther to the right. In the figure, $A$ and $B$ are on the same level line, so they land on the same spot, $z = 0$.

````python
w_1row = np.array([2, 1])
b_1row = -2
points_A_to_E = np.array([[1, 0], [0, 2], [2, 1], [-1, 1], [0.5, 2]])
print('scores z:', points_A_to_E @ w_1row + b_1row)
plot_plane_to_line(w_1row, b_1row, points_A_to_E);
````

### Step 2: two rows, from the plane to a new plane

Now stack two weight vectors, $\mathbf{w}_1 = (w_{11}, w_{12})$ and $\mathbf{w}_2 = (w_{21}, w_{22})$, as the rows of a matrix $W$, and their intercepts $b_1$ and $b_2$ as a vector $\mathbf{b}$:

$$
W = \begin{bmatrix} w_{11} & w_{12} \\ w_{21} & w_{22} \end{bmatrix} \quad \leftarrow \text{row 1 is } \mathbf{w}_1, \text{ row 2 is } \mathbf{w}_2, \qquad \mathbf{b} = \begin{bmatrix} b_1 \\ b_2 \end{bmatrix}.
$$

Each point $\mathbf{x}$ (written as a column) now lands at a new **point** $\mathbf{z} = (z_1, z_2)$:

$$
\mathbf{z} = W\mathbf{x} + \mathbf{b} = \begin{bmatrix} w_{11}x_1 + w_{12}x_2 + b_1 \\ w_{21}x_1 + w_{22}x_2 + b_2 \end{bmatrix} = \begin{bmatrix} \mathbf{w}_1\cdot\mathbf{x} + b_1 \\ \mathbf{w}_2\cdot\mathbf{x} + b_2 \end{bmatrix}.
$$

So $W\mathbf{x} + \mathbf{b}$ maps the feature plane to a new, two-dimensional z-space. **Each new coordinate is one row's score.** This is the same row view as in Part 6, except that now the rows of the matrix are weight vectors instead of data points, and each is dotted with the one point $\mathbf{x}$.

**Reading the subscripts.** $w_{kj}$ is in **row $k$** (which new coordinate) and **column $j$** (which feature it multiplies). So $w_{21}$ is row 2's weight on the first feature.

**An example: flattening the plane onto a line.** Take

$$
W = \begin{bmatrix} 1 & 1 \\ 2 & 2 \end{bmatrix}, \qquad \mathbf{b} = \begin{bmatrix} 0 \\ 0 \end{bmatrix}.
$$

Then $z_1 = x_1 + x_2$ and $z_2 = 2x_1 + 2x_2 = 2z_1$. The two rows point the same way (row 2 is twice row 1), so every point lands on the single line $z_2 = 2z_1$ in z-space. The figure shows what that does to five points:

- **Every point lands on one line.** The whole feature plane is flattened onto it.
- **Points that were far apart can land on top of each other.** $A = (2, 0)$ and $B = (-1, 3)$ are more than 4 units apart in feature space, but both have $x_1 + x_2 = 2$, so both land at $(2, 4)$.
- **Only one thing about each point survives:** its value of $x_1 + x_2$. Everything else about where it was is lost.

Part 9 explains which matrices flatten the plane like this, and why.

````python
W_step2 = np.array([[1, 1],
                    [2, 2]])     # row 2 is twice row 1
b_step2 = np.array([0, 0])
X_step2 = np.array([[2, 0], [-1, 3], [0, 0], [-2, 1], [1, 0.5]])   # points A to E
for name, x in zip('ABCDE', X_step2):
    print(name, x, 'lands at', W_step2 @ x + b_step2)
plot_plane_to_plane(W_step2, b_step2, X_step2);
````

## Part 8: Mapping many points at once: `X @ W.T + b`

So far:

- **Part 6: many points, one weight vector.** The data matrix $X$ holds the points as its rows. `X @ w + b` dots every point with the single weight vector $\mathbf{w}$, giving one score per point.
- **Part 7: one point, many weight vectors.** The weight matrix $W$ holds weight vectors as its rows. $W\mathbf{x} + \mathbf{b}$ dots the single point $\mathbf{x}$ with every weight vector, giving that point one new coordinate per row of $W$.

Now do both at once: dot **every point** (every row of $X$) with **every weight vector** (every row of $W$), so each point gets all of its new coordinates.

**The problem: both matrices store their vectors as rows.** A matrix product dots each **row** of the left matrix with each **column** of the right one (Part 6's shapes rule). $X$'s rows are the points, so $X$ belongs on the left. But the weight vectors are the **rows** of $W$, not its columns, so `X @ W` would dot each point with the wrong vectors.

**The fix: the transpose.** $W^T$ (read "$W$ transpose") swaps rows and columns: row $k$ of $W$ becomes column $k$ of $W^T$. In NumPy, it's `W.T`. With two weight vectors,

$$
W = \begin{bmatrix} w_{11} & w_{12} \\ w_{21} & w_{22} \end{bmatrix} \;\leftarrow \text{rows are } \mathbf{w}_1, \mathbf{w}_2,
\qquad
W^T = \begin{bmatrix} w_{11} & w_{21} \\ w_{12} & w_{22} \end{bmatrix} \;\leftarrow \text{columns are } \mathbf{w}_1, \mathbf{w}_2.
$$

So $W^T$ holds the weight vectors as **columns**, side by side, each one laid out like Part 6's column $\mathbf{w}$.

**Multiply.** $XW^T$ dots every row of $X$ with every column of $W^T$: every point with every weight vector. It's Part 6's $X\mathbf{w}$, done for all the weight vectors at once: column $k$ of $XW^T$ is $X\mathbf{w}_k$. The inner dimensions match because each point and each weight vector has $p$ entries:

$$
\underset{n \times p}{X}\;\;\underset{p \times k}{W^T} \;=\; \underset{n \times k}{XW^T}.
$$

Then add the intercepts:

$$
Z = X W^T + \mathbf{b}^T, \qquad \text{in code:} \quad \texttt{Z = X @ W.T + b}.
$$

$Z$ has **one row per point** and **one column per row of $W$** (one per new coordinate): `Z[i, k]` is point $i$'s $k$-th new coordinate, its score for row $k$ of $W$.

- **`W` is stored exactly as in the math**, one row per weight vector. Only the product needs those rows as columns, and `.T` provides that.
- **Broadcasting:** `b` has one entry per row of $W$ (shape `(k,)`). NumPy adds it to **every row** of `X @ W.T`, so each point gets each row's intercept. (In the formula, $\mathbf{b}^T$ is $\mathbf{b}$ laid out as a row.)

(You'll see exactly this formula again: PyTorch's standard layer computes `x @ W.T + b`, with `W` stored one row per output.)

<details>
<summary>Optional: the same formula from W x + b (the way Monday's lesson derives it)</summary>

Monday's lesson reaches `X @ W.T + b` from the other direction: it starts from Part 7's one-point formula $\mathbf{z} = W\mathbf{x} + \mathbf{b}$, where the point $\mathbf{x}$ is a column, and uses one rule of matrix algebra: **transposing a product reverses its order**, $(AB)^T = B^T A^T$.

Transpose both sides of $\mathbf{z} = W\mathbf{x} + \mathbf{b}$:

$$
\mathbf{z}^T = (W\mathbf{x})^T + \mathbf{b}^T = \mathbf{x}^T W^T + \mathbf{b}^T.
$$

Now the point $\mathbf{x}^T$ is a **row** going in, and its new coordinates $\mathbf{z}^T$ are a **row** coming out: the same numbers, laid out as rows. Since $X$ holds the points as rows, stacking every point's version of that equation gives $Z = XW^T + \mathbf{b}^T$, the same formula as above.

</details>

## Wrap-up

- **Vectors** can be drawn as points or arrows. $\mathbf{q} - \mathbf{p}$ is the vector from $\mathbf{p}$ to $\mathbf{q}$, and $\lVert \mathbf{x} \rVert$ is length.
- **The dot product** $\mathbf{w}\cdot\mathbf{x}$ is a weighted sum. Its sign gives the angle, and it's 0 exactly for perpendicular vectors.
- **$\mathbf{w}\cdot\mathbf{x} = c$** is a line perpendicular to $\mathbf{w}$. Different $c$'s give parallel lines, and a bigger $c$ puts the line farther out in $\mathbf{w}$'s direction.
- **$\mathbf{w}\cdot\mathbf{x} + b = 0$** is the same line with $c = -b$. $b$ slides it, and **$\mathbf{w}$ points to the side where the score is positive**. For logistic regression, $\boldsymbol\beta$ points toward class 1, perpendicular to the boundary.
- **A score is a height** whose level lines are those parallel lines; $\mathbf{w}$ points uphill.
- **`X @ w + b`** scores every point at once.
- **$W\mathbf{x} + \mathbf{b}$** maps a point to a new point: one row of $W$ gives a number line, two rows give a new plane. Each new coordinate is one row's score, and $w_{kj}$ is row $k$'s weight on feature $j$.
- **`X @ W.T + b`** maps every point at once: one row per point, one column per row of $W$ (one per new coordinate). The `.T` turns $W$'s rows (the weight vectors) into columns, so every point is dotted with every weight vector.

## Part 9: The column view: where can $W\mathbf{x}$ land?

Part 7 read $W\mathbf{x}$ **by rows**: each entry is a row of $W$ dotted with $\mathbf{x}$. There's a second way to read the same product, **by columns**. Multiply out a $2 \times 2$ matrix times a vector, and group the terms by $x_1$ and $x_2$:

$$
\begin{bmatrix} w_{11} & w_{12} \\ w_{21} & w_{22} \end{bmatrix}\begin{bmatrix} x_1 \\ x_2 \end{bmatrix}
= \begin{bmatrix} w_{11} x_1 + w_{12} x_2 \\ w_{21} x_1 + w_{22} x_2 \end{bmatrix}
= x_1 \begin{bmatrix} w_{11} \\ w_{21} \end{bmatrix} + x_2 \begin{bmatrix} w_{12} \\ w_{22} \end{bmatrix}.
$$

So $W\mathbf{x}$ is always **$x_1$ times the first column, plus $x_2$ times the second column**. Whatever $\mathbf{x}$ is, the result is a combination of $W$'s columns.

**The column space.** The set of all such combinations is called the **column space** of $W$: every place $W\mathbf{x}$ can possibly land.

- If the two columns point in **different directions**, their combinations reach every point of the plane. The column space is the whole plane.
- If the two columns are **parallel** (one is a multiple of the other, or one is zero), every combination is a multiple of that one direction. The column space is just a **line through the origin**. The matrix **flattens** the whole plane onto that line.
- Adding $\mathbf{b}$ shifts everything by $\mathbf{b}$, so $W\mathbf{x} + \mathbf{b}$ lands on that line moved by $\mathbf{b}$.

**The flip side: many inputs land on one spot.** When the plane is flattened onto a line, lots of inputs must land on each output point. Which ones? For example, with both columns equal to $(1, 1)$,

$$
\begin{bmatrix} 1 & 1 \\ 1 & 1 \end{bmatrix}\mathbf{x} = x_1\begin{bmatrix} 1 \\ 1 \end{bmatrix} + x_2\begin{bmatrix} 1 \\ 1 \end{bmatrix} = (x_1 + x_2)\begin{bmatrix} 1 \\ 1 \end{bmatrix}.
$$

The output depends only on $x_1 + x_2$, which is the dot product $(1, 1)\cdot\mathbf{x}$. All the inputs with the same value of $x_1 + x_2$, a whole line from Part 3, land on the **same output point**.

## Part 10: A matrix moves the plane

Apply $\mathbf{x} \mapsto W\mathbf{x} + \mathbf{b}$ to **every** point of the plane at once, and watch what happens to a square grid. Three facts:

1. **Straight lines stay straight, and parallel lines stay parallel and evenly spaced.** A line is $\mathbf{p} + t\mathbf{d}$ for numbers $t$. Its image is $W(\mathbf{p} + t\mathbf{d}) + \mathbf{b} = (W\mathbf{p} + \mathbf{b}) + t\,(W\mathbf{d})$: again a line (or a single point, if $W\mathbf{d} = \mathbf{0}$). Parallel lines share their $\mathbf{d}$, so their images share $W\mathbf{d}$ and stay parallel. Equal steps in $t$ map to equal steps, so even spacing survives too.
2. **The columns say where the grid goes.** $(1, 0)$ lands on column 1 of $W$ and $(0, 1)$ on column 2 (the column view, Part 9). The whole grid follows those two vectors, so the unit square becomes the parallelogram they span.
3. **$\mathbf{b}$ shifts everything.** $W\mathbf{0} = \mathbf{0}$, so without $\mathbf{b}$ the origin stays put. Adding $\mathbf{b}$ moves the whole picture by $\mathbf{b}$.

So a matrix plus a shift can **stretch, flip, rotate, shear, flatten, and shift** the plane, but every straight line stays straight.

## Part 11: Two maps in a row are one map *(read and run)*

Apply one map, then another: first $\mathbf{y} = W_1\mathbf{x} + \mathbf{b}_1$, then $W_2\mathbf{y} + \mathbf{b}_2$. Substitute and multiply out:

$$
W_2(W_1\mathbf{x} + \mathbf{b}_1) + \mathbf{b}_2 = \underbrace{(W_2W_1)}_{\text{one matrix}}\mathbf{x} + \underbrace{(W_2\mathbf{b}_1 + \mathbf{b}_2)}_{\text{one shift}}.
$$

**Two maps in a row are always a single map of the same kind,** with matrix $W_2W_1$ and shift $W_2\mathbf{b}_1 + \mathbf{b}_2$. This is what matrix multiplication *is*: doing one map, then another. (Note the order: $W_1$ acts first, but it's written on the right.)

Shapes compose the same way. A $3 \times 2$ matrix (2 numbers in, 3 out) followed by a $1 \times 3$ matrix (3 in, 1 out) is a single $1 \times 2$ matrix (2 in, 1 out).

**Optional viewing:** 3Blue1Brown's [*Matrix multiplication as composition*](https://www.3blue1brown.com/lessons/matrix-multiplication).

**Code worth reading.** One pair of maps, combined, and checked on a batch of points.

````python
W1_demo = np.array([[1, 2], [0, 1]])
b1_demo = np.array([1, 0])
W2_demo = np.array([[3, -1], [1, 1]])
b2_demo = np.array([0, 2])

W_comb_demo = W2_demo @ W1_demo                     # one matrix
b_comb_demo = W2_demo @ b1_demo + b2_demo           # one shift

pts_demo = np.random.default_rng(1).normal(size=(6, 2))          # six points, one per row
two_steps = (pts_demo @ W1_demo.T + b1_demo) @ W2_demo.T + b2_demo
one_step = pts_demo @ W_comb_demo.T + b_comb_demo
print('combined W:', W_comb_demo.tolist(), ' combined b:', b_comb_demo)
print('two steps == one step?', np.allclose(two_steps, one_step))
````

### O1 · From log-odds to probability

Logistic regression models the log-odds as a score $z$: $\ln\!\left(\frac{p}{1-p}\right) = z$. Exponentiate both sides to get $\frac{p}{1-p} = e^{z}$, then solve for $p$: $p = e^z(1 - p)$, so $p(1 + e^z) = e^z$, and

$$
p = \frac{e^z}{1 + e^z} = \frac{1}{1 + e^{-z}} = \sigma(z).
$$

**Try it:** what is $p$ when $z = 0$? When $z = 2$? Why is $p > 0.5$ exactly when $z > 0$?

<details>
<summary>Answer</summary>

$\sigma(0) = 1/2$, and $\sigma(2) = 1/(1 + e^{-2}) \approx 0.881$. $\sigma$ is increasing and equals $0.5$ at $z = 0$, so it's above 0.5 exactly when $z > 0$.

</details>

### O4 · Which points map onto a line?

Suppose a map sends $\mathbf{x}$ to $\mathbf{z} = W\mathbf{x} + \mathbf{b}$, and in the new plane you draw a line $\mathbf{u}\cdot\mathbf{z} + c = 0$. Which original points $\mathbf{x}$ land on it?

**Try it:** substitute $\mathbf{z} = W\mathbf{x} + \mathbf{b}$, and rewrite the condition in the form $\mathbf{v}\cdot\mathbf{x} + d = 0$. (Use the fact that $\mathbf{u}\cdot(W\mathbf{x}) = (W^T\mathbf{u})\cdot\mathbf{x}$, which is Part 8's transpose rule.) What kind of set is it?

<details>
<summary>Answer</summary>

$\mathbf{u}\cdot(W\mathbf{x} + \mathbf{b}) + c = (W^T\mathbf{u})\cdot\mathbf{x} + (\mathbf{u}\cdot\mathbf{b} + c) = 0$. That's a line in the original plane, with $\mathbf{v} = W^T\mathbf{u}$ and $d = \mathbf{u}\cdot\mathbf{b} + c$ (unless $W^T\mathbf{u} = \mathbf{0}$, in which case it's every point or none). So **a line drawn after the map is a line before the map**: one straight-line split of the new plane is a straight-line split of the original.

</details>

### O5 · Distance from a line

A point's distance from the line $\mathbf{w}\cdot\mathbf{x} + b = 0$ is $\lvert \mathbf{w}\cdot\mathbf{x} + b \rvert / \lVert \mathbf{w} \rVert$. Here's why: the closest point of the line is straight across, in $\mathbf{w}$'s direction, at $\mathbf{x} - t\mathbf{w}$ for some $t$. Setting its score to 0 gives $(\mathbf{w}\cdot\mathbf{x} + b) - t\lVert \mathbf{w} \rVert^2 = 0$, so $t = (\mathbf{w}\cdot\mathbf{x} + b)/\lVert \mathbf{w} \rVert^2$, and the distance is $\lvert t \rvert\,\lVert \mathbf{w} \rVert$.

**Try it:** with $\mathbf{w} = (1, 1)$ and $b = -2$, how far is $(3, 0)$ from the line? Which point of the line is closest to it?

<details>
<summary>Answer</summary>

The score is $3 + 0 - 2 = 1$, so the distance is $1/\sqrt2 \approx 0.71$. $t = 1/2$, so the closest point is $(3, 0) - 0.5\,(1, 1) = (2.5, -0.5)$. Check: $2.5 - 0.5 - 2 = 0$. ✓

</details>
